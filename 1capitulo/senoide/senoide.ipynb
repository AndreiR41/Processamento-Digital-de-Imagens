```python
from PIL import Image
import yaml
import numpy as np
import matplotlib.pyplot as plt
```

```python
from google.colab import drive
drive.mount('/content/drive')

yml_file_path = '/content/drive/MyDrive/PDI/1capitulo/senoide/build/senoide-256.yml'
png_file_path = '/content/drive/MyDrive/PDI/1capitulo/senoide/build/senoide-256.png'
# tive que editar o arquivo para poder abrir
with open(yml_file_path, 'r') as file:
    yml_data = yaml.load(file, Loader=yaml.FullLoader)


img = Image.open(png_file_path)
png_matrix = np.array(img)

# Convert the YAML data to a NumPy matrix
yml_matrix = np.array(yml_data, dtype=np.float32).reshape(256, 256)
print("YAML matrix shape:", yml_matrix.shape)
print("PNG matrix shape:", png_matrix.shape)
```

```python
sen_float = yml_matrix[0]
sen_int = png_matrix[0]

plt.figure(figsize=(12, 4))

# Primeiro gráfico - dados float do YAML
plt.subplot(1, 2, 1)
plt.plot(sen_float)
plt.title('YAML - Primeira linha (float)')
plt.xlabel('Coluna')
plt.ylabel('Valor')
plt.grid(True)

# Segundo gráfico - dados int do PNG
plt.subplot(1, 2, 2)
plt.plot(sen_int)
plt.title('PNG - Primeira linha (int)')
plt.xlabel('Coluna')
plt.ylabel('Valor')
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
sen_dif = sen_float - sen_int
plt.plot(sen_dif)
plt.title('Diferença entre os dados')
plt.xlabel('Coluna')
plt.ylabel('Valor')
plt.grid(True)
plt.show()
```

```python
# Converter int para float
plt.figure(figsize=(10, 10))

plt.plot(sen_float, label='YAML (float original)', linewidth=8)
plt.plot(sen_int, label='PNG (int convertido para float)', linewidth=2, alpha=0.8)

plt.title('Comparação: YAML vs PNG - Primeira linha', fontsize=14)
plt.xlabel('Coluna', fontsize=12)
plt.ylabel('Valor', fontsize=12)
plt.legend(fontsize=12)
plt.grid(True, alpha=0.3)
plt.show()
```
